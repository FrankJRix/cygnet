## From mechanics to a physics result

A trained checkpoint and matched preprocessing are required for meaningful predictions. The thesis evaluates missed simulated signal counts and retained pixels, rather than treating noisy camera sums as deposited energy.

On 900 simulated test frames the thesis reports mean missed energy (0.12 ± 0.11)% and mean pixel retention (0.38 ± 0.02)%, with uncertainties on the means. One weak event is missed entirely. Real-data applications are qualitative.
